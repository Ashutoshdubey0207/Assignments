Q6. What do you understand about the windows function in pandas and list the types of windows
functions?

Window functions in pandas perform calculations across a sliding or cumulative set of rows, returning a value for every original row in the dataset

```python
# Rolling windows
df2 = pd.DataFrame({
    "Values" : [10,20,30,40,50]
})
```

```python
df2["rolling_mean"] = df2["Values"].rolling(window=2).mean()
```

```python
df2
```

```python
# Expanding Window
df2["Expanding_sum"] = df2["Values"].expanding(min_periods=2).sum()
```

```python
df2
```

```python
## Exponentially weighted window
df2["Eww_mean"] = df2['Values'].ewm(span=2,adjust=False).mean()
```

```python
df2
```

```python
## Weighted Window
df2['weighted_window'] = df2['Values'].rolling(window=3,win_type='triang').mean()
```

```python
df2
```

Q9. Write a Python program that reads a CSV file containing categorical data and converts a specified
column to a categorical data type. The program should prompt the user to enter the file path, column
name, and category order, and then display the sorted data.

```python
file_path = input("enter file path : ")
column_name = input("enter column name: ")
category_order = input("enter category order (comma seperated)").split(",")
# Remove the extra space from the category order
category_order = [category.strip() for category in category_order]

df = pd.read_csv(file_path)

df[column_name] = pd.Categorical(
    df[column_name],categories=category_order,ordered=True
)
df_sorted = df.sort_values(by=column_name)
print("\n Sorted data :")
df_sorted
```

Q10. Write a Python program that reads a CSV file containing sales data for different products and
visualizes the data using a stacked bar chart to show the sales of each product category over time. The
program should prompt the user to enter the file path and display the chart.

```python
import matplotlib.pyplot as plt
file_path = input("Enter file path : ")
df = pd.read_csv(file_path, parse_dates=["Date"])
df["Quarter"] =df["Date"].dt.to_period("Q").astype(str)
pivot = df.pivot_table(index="Quarter",columns="Category",values="Units_Sold")
pivot.plot(kind="bar",figsize=(10,5))
plt.ylabel("Unit sold")
plt.title("Quaretly units sold by category")
plt.xticks(rotation=45)
plt.tight_layout()
```

Q11. You are given a CSV file containing student data that includes the student ID and their test score. Write
